```python
import sqlite3
```

```python
conn = sqlite3.connect('../metacyc/feba.db')
cursor = conn.cursor()
```

```python
prefix = "Metacyc"
```

```python
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()
```

```python
tables
```

```python
cursor.execute("SELECT * FROM MetacycPathwayReaction LIMIT 10;")
```

```python
cursor.fetchall()
```

```python
# Filter and drop tables
for table in tables:
    table_name = table[0]
    if not table_name.startswith(prefix):
        cursor.execute(f"DROP TABLE {table_name}")
        print(f"Dropped table {table_name}")

# Commit changes and close connection
conn.commit()
conn.close()
```
